```python
df=df_train.merge(df_test,how='outer')
```

```python
#statistical summery
```

```python
df.describe()
```

```python
#We can drop the USER_ID column
```

```python
df.drop(['User_ID'],axis=1,inplace=True)
```

```python
df.head()
```

```python
#Here we can change the Gender (categorical) to 0 or 1
```

```python
df['Gender']=df['Gender'].map({'F':0,'M':1})
```

```python
df.head(5)
```

```python
#Convert AGE(continuous categorical) variable in to numerical.
```

```python
df['Age'].unique()
```

```python
#we can map here to convert age to numerical
```

```python
df['Age']=df['Age'].map({'0-17':1,'18-25':2, '26-35':3, '36-45':4, '46-50':5, '51-55':6,'55+':7})
```

```python
df.head()
```

```python
#Now we can convert the city, which is categorical in to numeric.
```

```python
df['City_Category'].unique()
```

```python
df_city=pd.get_dummies(df['City_Category'],drop_first=True).astype(int)
```

```python
df_city.head()
```

```python
#concat the above encoded city in to original df
```

```python
df=pd.concat([df,df_city],axis=1)
```

```python
df.drop('City_Category',axis=1,inplace=True)
```

```python
df.head(5)
```

```python
#find sum of missing values
```

```python
df.isnull().sum()
```

```python
#we can fill product_category 2 with mode value
```

```python
df['Product_Category_2'].value_counts()
```

```python
df['Product_Category_2'].mode()[0]
```

```python
df['Product_Category_2']=df['Product_Category_2'].fillna(df['Product_Category_2'].mode()[0])
```

```python
df['Product_Category_2'].isnull().sum()
```

```python
#similarly we can fill mode value in product category 3
```

```python
df['Product_Category_3'].value_counts()
```

```python
df['Product_Category_3']=df['Product_Category_3'].fillna(df['Product_Category_3'].mode()[0])
```

```python
df['Product_Category_3'].isnull().sum()
```

```python
df.info()
```

```python
#Here Stay_in_current_city is a string.we have to change to int,but first we have to replace + in the column before using astype(int)
```

```python
df['Stay_In_Current_City_Years']=df['Stay_In_Current_City_Years'].str.replace('+','')
```

```python
df['Stay_In_Current_City_Years']=df['Stay_In_Current_City_Years'].astype(int)
```
